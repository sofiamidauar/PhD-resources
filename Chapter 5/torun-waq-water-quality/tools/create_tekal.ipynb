```python
import pandas as pd
```

### Write profile data 

```python
# Load CSV
df = pd.read_csv("monthly_measurements_CAB.csv", encoding='latin1')

# Convert Date to YYYYMMDD
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True).dt.strftime('%Y%m%d').astype(int)

# Convert Heure to HHMMSS
df['Time'] = pd.to_datetime(df['Heure'], format='%H:%M').dt.strftime('%H%M%S').astype(int)

# Pivot data: rows = Date+Time, columns = Depth, values = Con O2
df_pivot = df.pivot_table(index=['Date', 'Time'], columns='Depth (m)', values='Con O2 (mg/L)')

# Sort depths
df_pivot = df_pivot.sort_index(axis=1)

# Fill NaN values with 0
df_pivot = df_pivot.fillna(0)

# Tekal parameters
column_lines = [
    "* column 1 : date",
    "* column 2 : time"
] + [f"* column {i+3} : DO {int(d)}m CAB" for i, d in enumerate(df_pivot.columns)]

num_rows = df_pivot.shape[0]
num_cols = df_pivot.shape[1] + 2  # Date + Time + depth columns

column_names = ['Date', 'Time'] + [f"D{int(d)}" for d in df_pivot.columns]

output_file = "ConO2.tek"

with open(output_file, "w") as f:
    # Write comments
    for line in column_lines:
        f.write(line + "\n")
    
    # Number of rows and columns
    f.write(f"Measured profiles DO CAB\n")
    f.write(f"{num_rows} {num_cols}\n")
    
    # Write data lines
    for (date, time), row in df_pivot.iterrows():
        line_str = f"{date} {time:06d}"
        for val in row:
            line_str += f" {val:05.2f}"  # fixed width: always 5 chars, 2 decimals
        f.write(line_str + "\n")
```

### Writing just one depth - high freq data

```python
# Step 1: Load your CSV file
df = pd.read_csv("data/DO_CAB_format_totekal.csv  # was an absolute OneDrive path; place file under data/", index_col=False)
# Remove any extra unnamed columns (like indices)
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]

# Step 2: Tekal file settings
column_names = ['date', 'time', 'DO']  # Your 3 columns
comment_lines = [
    "* column 1 : date",
    "* column 2 : time",
    "* column 3 : DO 0.3m CAB",
    "Measured DO CAB"
]

output_file = "DO_measured_CAB.tek"

# Step 3: Write Tekal file
with open(output_file, "w") as f:
    # Write comment lines
    for line in comment_lines:
        f.write(line + "\n")
    
    # Write number of rows and columns
    num_rows = len(df)
    num_cols = len(column_names)
    f.write(f"{num_rows} {num_cols}\n")
    
    # Write data lines
    for _, row in df.iterrows():
        # Format Date as int, Time as 6-digit zero-padded, DO with 2 decimals
        line_str = f"{int(row['date'])} {int(row['time']):06d} {row['DO']:.2f}"
        f.write(line_str + "\n")
```
